```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("GoToCompany/llama3-8b-cpt-sahabatai-v1-instruct")

print("Pad token:", tokenizer.pad_token, "ID:", tokenizer.pad_token_id)
print("Eos token:", tokenizer.eos_token, "ID:", tokenizer.eos_token_id)
print("Vocab size:", tokenizer.vocab_size)
```

```python
print("Token untuk ID 0:", tokenizer.convert_ids_to_tokens(0))
```

```python
print(f"Token untuk ID {tokenizer.vocab_size}:", tokenizer.convert_ids_to_tokens(tokenizer.vocab_size))
```

```python
print("Token untuk ID 128001:", tokenizer.convert_ids_to_tokens(128001))
```

```python
tokenizer.add_special_tokens({"additional_special_tokens": ["[BLANK]"]})
blank_id = tokenizer.convert_tokens_to_ids("[BLANK]")

print("Blank token:", "[BLANK]")
print("Blank ID:", blank_id)
print("Total vocab size:", tokenizer.vocab_size)
```

```python
if "[BLANK]" not in tokenizer.get_vocab():
    tokenizer.add_special_tokens({"additional_special_tokens": ["[BLANK]"]})
    print(f"✅ Token [BLANK] ditambahkan, total vocab sekarang: {len(tokenizer)}")
```
